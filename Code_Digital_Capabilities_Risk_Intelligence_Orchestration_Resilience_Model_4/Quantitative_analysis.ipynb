```python

from scipy.stats import f, ncf

# A priori power analysis settings
alpha = 0.05
target_power = 0.90
effect_size_f2 = 0.15
number_of_predictors = 3
actual_sample_size = 409

def calculate_power(n, predictors, f2, alpha=0.05):
    """Power for multiple regression: R² deviation from zero."""
    df1 = predictors
    df2 = n - predictors - 1

    if df2 <= 0:
        return 0.0

    critical_f = f.isf(alpha, df1, df2)
    noncentrality = f2 * n

    return ncf.sf(
        critical_f,
        df1,
        df2,
        noncentrality
    )

# Find the minimum required sample size
required_n = None

for n in range(number_of_predictors + 2, 10000):
    achieved_power = calculate_power(
        n,
        number_of_predictors,
        effect_size_f2,
        alpha
    )

    if achieved_power >= target_power:
        required_n = n
        break

# Calculate power for the actual sample size
actual_power = calculate_power(
    actual_sample_size,
    number_of_predictors,
    effect_size_f2,
    alpha
)

print("A PRIORI POWER ANALYSIS")
print("-" * 40)
print(f"Effect size (f²): {effect_size_f2}")
print(f"Significance level (alpha): {alpha}")
print(f"Target statistical power: {target_power}")
print(f"Number of predictors: {number_of_predictors}")
print(f"Required sample size: {required_n}")
print(f"Actual sample size: {actual_sample_size}")
print(f"Power at actual sample size: {actual_power:.4f}")
print(f"Power at actual sample size (%): {actual_power * 100:.2f}%")
print(f"Sufficient for target power: {actual_power >= target_power}")
```
