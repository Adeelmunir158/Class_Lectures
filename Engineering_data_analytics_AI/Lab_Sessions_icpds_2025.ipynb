# Lab session 1
A batch of 10 steel rods is tested for tensile strength. The minimum acceptable tensile strength is 250 MPa. Write a Python program using a for loop and if/else statements that:

Reads tensile strength values for 10 rods.

Prints whether each rod is “Pass” or “Fail” based on the criterion.

Counts and displays the total number of rods that passed and failed.

# Lab Session 2
In a mechanical workshop, students are asked to record the diameter (in mm) of 5 manufactured shafts. The acceptable tolerance range is 49 mm to 51 mm.

Write a Python program that:

Uses a for loop to repeatedly take input from the user for the diameters of 5 shafts.

For each entered diameter, checks whether it lies within the acceptable tolerance range.

If yes → print “Within Tolerance”

If no → print “Out of Tolerance”

At the end of the loop, display the total number of shafts that passed and failed.

# Lab Session 4

In a mechanical workshop, students are asked to evaluate the diameter (in mm) of 12 manufactured shafts. The acceptable tolerance range is 49 mm to 51 mm (inclusive). Use the NumPy library extensively and avoid for/while loops.

Write a Python program that:

Imports NumPy (import numpy as np).

Takes a single line of input containing 12 diameters separated by spaces (e.g., 49.8 50.1 51.2 ...) and converts it to a NumPy array

Creates a boolean mask with NumPy to check tolerance for all shafts at once: (d >= 49) & (d <= 51).

Uses np.where to generate a vector of labels: “Within Tolerance” / “Out of Tolerance” for each shaft (no explicit loops).

Displays (using NumPy functions only):

Total shafts tested 

Number within tolerance 

Number out of tolerance 

Indices of out-of-tolerance shafts 

Prints basic statistics of the measured diameters using NumPy: mean, standard deviation (sample: ddof=1), min, and max.
